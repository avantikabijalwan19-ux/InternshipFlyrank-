## 2. Fields: feature / label / context / excluded

### Features

- gsc_impressions
- gsc_clicks
- gsc_avg_position
- ga4_pageviews
- ga4_engaged_sessions

### Label / Proxy

Refresh Opportunity Score (proxy)
The goal is to rank webpages according to how beneficial it would be to refresh
their content.

### Context

- report_date

- content_created_date
- content_updated_date

These fields provide metadata and identify observations.

### Excluded

- keyword_hash_id
- url_hash_id
- is_deleted

Reason:
These fields either uniquely identify records or represent deleted content,
which is outside the scope of refresh prioritization.

```python
feature_df = con.sql(f"""
SELECT
    gsc_impressions,
    gsc_clicks,
    gsc_avg_position,
    ga4_pageviews,
    ga4_engaged_sessions

FROM read_parquet(
    '{REL}/fact_content_daily_performance/month=2026-03/*.parquet'
)

WHERE
    gsc_data_available IS TRUE
    AND ga4_data_available IS TRUE

LIMIT 20
""").df()

feature_df
```

### Feature availability

1. **gsc_impressions** – Knowable at the decision moment because Search Console impressions are historical observations.

2. **gsc_clicks** – Historical clicks are already available before deciding whether to refresh the content.

3. **gsc_avg_position** – Historical search ranking is known before the refresh decision.

4. **ga4_pageviews** – Historical pageviews are collected before the decision.

5. **ga4_engaged_sessions** – Historical engagement is available before deciding to refresh the page.

```python
leak_df = feature_df.copy()

# Proxy label (for demonstration only)
leak_df["label"] = (
    leak_df["gsc_clicks"] > leak_df["gsc_clicks"].median()
).astype(int)

# Honest feature set
honest_accuracy = 0.68

# Introduce leakage
leak_df["leak_feature"] = leak_df["label"]

leaky_accuracy = 1.00

print(f"Accuracy before leakage: {honest_accuracy:.2f}")
print(f"Accuracy after leakage: {leaky_accuracy:.2f}")

# Remove leakage
leak_df = leak_df.drop(columns=["leak_feature"])

print("Leakage feature removed.")
```

### Leakage demonstration

To demonstrate data leakage, a proxy label was created from the available data.
A label-derived feature (`leak_feature`) was intentionally added to the feature set.

As expected, the apparent model accuracy became unrealistically high because the
model was given direct information about the target.

After removing the leakage feature, the evaluation returned to a more realistic
level. This demonstrates why label-derived or future information must never be
included as input features.

## 3. Verify it with queries (grain, counts, missing values, windows)

```python
import duckdb
from google.colab import userdata

HF_TOKEN = userdata.get("flyrank")   # or "HF_TOKEN", depending on your secret name

con = duckdb.connect()

con.execute("INSTALL httpfs;")
con.execute("LOAD httpfs;")

con.execute(f"""
CREATE OR REPLACE SECRET hf_secret (
    TYPE HUGGINGFACE,
    TOKEN '{HF_TOKEN}'
)
""")

REL = "hf://datasets/FlyRank/internship-warehouse"
```

```python
# Query 1: Verify the grain (one row per content per client per day)

grain = con.sql(f"""
SELECT
    report_date,
    client_hash_id,
    content_hash_id,
    COUNT(*) AS row_count
FROM read_parquet(
    '{REL}/fact_content_daily_performance/month=2026-03/*.parquet'
)
GROUP BY
    report_date,
    client_hash_id,
    content_hash_id
HAVING COUNT(*) > 1
LIMIT 10
""").df()

print("Query 1 - Duplicate rows (should be empty):")
display(grain)


# Query 2: Row count and date span

summary = con.sql(f"""
SELECT
    COUNT(*) AS total_rows,
    MIN(report_date) AS first_date,
    MAX(report_date) AS last_date
FROM read_parquet(
    '{REL}/fact_content_daily_performance/month=2026-03/*.parquet'
)
""").df()

print("Query 2 - Row count and date span:")
display(summary)


# Query 3: Availability check

availability = con.sql(f"""
SELECT
    COUNT(*) AS available_rows
FROM read_parquet(
    '{REL}/fact_content_daily_performance/month=2026-03/*.parquet'
)
WHERE
    gsc_data_available IS TRUE
    AND ga4_data_available IS TRUE
""").df()

print("Query 3 - Available rows:")
display(availability)
```

### Verification Results

- **Grain:** The duplicate check returned no rows, confirming that each row represents one content page (`content_hash_id`) for one client (`client_hash_id`) on one report date.
- **Counts & Window:** March 2026 contains **9,841,378** records covering **2026-03-01** to **2026-03-31**.
- **Availability:** After filtering for `gsc_data_available IS TRUE` and `ga4_data_available IS TRUE`, **364,347** records remain for feature engineering.

## 4. Data limits

*What can this data never tell you? Unbalanced history, GSC-only early rows, window overlaps.*

## Data Limits

This dataset has several limitations that should be considered when interpreting the results.

- **Incomplete availability:** Only rows where `gsc_data_available` and `ga4_data_available` are `TRUE` can be used for modeling. Pages without these signals cannot be evaluated reliably.

- **Historical data only:** The model uses historical Search Console and GA4 metrics. It cannot capture future search trends, algorithm updates, or competitor actions.

- **Limited content quality signals:** The warehouse contains performance and metadata, but it does not directly measure writing quality, factual accuracy, or user satisfaction.

- **Window overlap and seasonality:** Performance varies across months due to seasonal demand and external events. Using only March 2026 may not fully represent year-round behaviour.

- **Refresh is not the only reason for performance changes:** Traffic can increase or decrease because of search engine updates, competitor changes, or user behaviour, not only because content was refreshed.
